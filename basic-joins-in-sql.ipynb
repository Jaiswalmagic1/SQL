```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import sqlite3
import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
```

```python
sqliteConnection = sqlite3.connect('ecommerce_db')
cursor = sqliteConnection.cursor()
```

```python
def execute(query):
    cursor.execute(query)
    result = cursor.fetchall()
    for r in result:
        print(r)
```

# Creating table

## Create table for Categories

```python
query = '''
CREATE TABLE Categories (
    CategoryID INTEGER PRIMARY KEY,
    CategoryName TEXT NOT NULL
);
'''
execute(query)
```

## Create table for Customers

```python
query = '''
CREATE TABLE Customers (
    CustomerID INTEGER PRIMARY KEY,
    FirstName TEXT NOT NULL,
    LastName TEXT NOT NULL,
    Email TEXT NOT NULL UNIQUE,
    Phone TEXT,
    Address TEXT,
    City TEXT,
    State TEXT,
    ZipCode TEXT,
    Country TEXT
);
'''
execute(query)
```

## Create table for Products

```python
query = '''
CREATE TABLE Products (
    ProductID INTEGER PRIMARY KEY,
    ProductName TEXT NOT NULL,
    Description TEXT,
    Price REAL NOT NULL,
    StockQuantity INTEGER NOT NULL,
    CategoryID INTEGER,
    FOREIGN KEY (CategoryID) REFERENCES Categories(CategoryID)
);

'''
execute(query)
```

## Create table for Orders

```python
query = '''
CREATE TABLE Orders (
    OrderID INTEGER PRIMARY KEY,
    CustomerID INTEGER,
    OrderDate TEXT NOT NULL,
    TotalAmount REAL NOT NULL,
    FOREIGN KEY (CustomerID) REFERENCES Customers(CustomerID)
);

'''
execute(query)
```

## Create table for OrderItems

```python
query = '''
CREATE TABLE OrderItems (
    OrderItemID INTEGER PRIMARY KEY,
    OrderID INTEGER,
    ProductID INTEGER,
    Quantity INTEGER NOT NULL,
    Price REAL NOT NULL,
    FOREIGN KEY (OrderID) REFERENCES Orders(OrderID),
    FOREIGN KEY (ProductID) REFERENCES Products(ProductID)
);

'''
execute(query)
```

## Create table for Payments

```python
query = '''
CREATE TABLE Payments (
    PaymentID INTEGER PRIMARY KEY,
    OrderID INTEGER,
    Amount REAL NOT NULL,
    PaymentMethod TEXT NOT NULL,
    PaymentStatus TEXT NOT NULL,
    FOREIGN KEY (OrderID) REFERENCES Orders(OrderID)
);
'''
execute(query)
```

## Create table for Reviews

```python
query = '''
CREATE TABLE Reviews (
    ReviewID INTEGER PRIMARY KEY,
    ProductID INTEGER,
    CustomerID INTEGER,
    Rating INTEGER NOT NULL CHECK(Rating BETWEEN 1 AND 5),
    ReviewText TEXT,
    FOREIGN KEY (ProductID) REFERENCES Products(ProductID),
    FOREIGN KEY (CustomerID) REFERENCES Customers(CustomerID)
);
'''
execute(query)
```

## Write a query to list all orders with customer information.


```python
query = '''
SELECT *
FROM orders o inner join Customers c on o.CustomerID = c.CustomerID;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Find all customers with no orders.


```python
query = '''
SELECT *
FROM Customers c left join orders o on o.CustomerID = c.CustomerID
where o.orderid is null;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## List products that have not been ordered.


```python
query = '''
SELECT p.ProductID, p.ProductName
FROM Products p left join OrderItems i on p.productid = i.productid
WHERE i.ProductID is null;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Show all orders and products, even if they don't match.

```python
query = '''
SELECT o.OrderID, o.ProductID, p.ProductName
FROM orderitems o left join products p on o.productid = p.productid

union

SELECT o.OrderID, o.ProductID, p.ProductName
FROM products p left join orderitems o on p.productid = o.productid
'''
df = pd.read_sql_query(query, sqliteConnection)
print('''Since SQLite does not support FULL OUTER JOIN this is a workaround we have used otherwise the FULL OUTER JOIN would have been as below.\n
SELECT o.OrderID, o.ProductID, p.ProductName\n
FROM OrderItems FULL OUTER JOIN Products p ON o.ProductID = p.ProductID;
''')
df
```
