# Creating table

## Patients Table

```python
query = '''
CREATE TABLE Patients (
    PatientID INTEGER PRIMARY KEY AUTOINCREMENT,
    FirstName TEXT NOT NULL,
    LastName TEXT NOT NULL,
    City TEXT NOT NULL,
    InsurancePlanID INTEGER,
    FOREIGN KEY (InsurancePlanID) REFERENCES InsurancePlans(InsurancePlanID)
);
'''
execute(query)
```

## Doctors Table

```python
query = '''
CREATE TABLE Doctors (
    DoctorID INTEGER PRIMARY KEY AUTOINCREMENT,
    FirstName TEXT NOT NULL,
    LastName TEXT NOT NULL,
    Specialty TEXT NOT NULL,
    HospitalID INTEGER,
    FOREIGN KEY (HospitalID) REFERENCES Hospitals(HospitalID)
);
'''
execute(query)
```

## Hospitals Table

```python
query = '''
CREATE TABLE Hospitals (
    HospitalID INTEGER PRIMARY KEY AUTOINCREMENT,
    HospitalName TEXT NOT NULL,
    City TEXT NOT NULL
);
'''
execute(query)
```

## Procedures Table

```python
query = '''
CREATE TABLE Procedures (
    ProcedureID INTEGER PRIMARY KEY AUTOINCREMENT,
    ProcedureName TEXT NOT NULL,
    Cost DECIMAL(10, 2) NOT NULL
);
'''
execute(query)
```

## InsurancePlans Table

```python
query = '''
CREATE TABLE InsurancePlans (
    InsurancePlanID INTEGER PRIMARY KEY AUTOINCREMENT,
    PlanName TEXT NOT NULL,
    InsuranceProvider TEXT NOT NULL
);
'''
execute(query)
```

## Consultations Table

```python
query = '''
CREATE TABLE Consultations (
    ConsultationID INTEGER PRIMARY KEY AUTOINCREMENT,
    PatientID INTEGER,
    DoctorID INTEGER,
    ProcedureID INTEGER,
    ConsultationDate DATE NOT NULL,
    FOREIGN KEY (PatientID) REFERENCES Patients(PatientID),
    FOREIGN KEY (DoctorID) REFERENCES Doctors(DoctorID),
    FOREIGN KEY (ProcedureID) REFERENCES Procedures(ProcedureID)
);
'''
execute(query)
```

## InsuranceClaims Table


```python
query = '''
CREATE TABLE InsuranceClaims (
    ClaimID INTEGER PRIMARY KEY AUTOINCREMENT,
    PatientID INTEGER,
    ProcedureID INTEGER,
    ClaimDate DATE NOT NULL,
    FOREIGN KEY (PatientID) REFERENCES Patients(PatientID),
    FOREIGN KEY (ProcedureID) REFERENCES Procedures(ProcedureID)
);
'''
execute(query)
```

```python
tables_query = "SELECT name FROM sqlite_master WHERE type='table';"
tables = pd.read_sql_query(tables_query, sqliteConnection)

for table in tables['name']:
    cursor = sqliteConnection.execute(f"PRAGMA table_info({table});")
    columns = cursor.fetchall()

    # Create a DataFrame for the current table's column info
    table_info_df = pd.DataFrame(columns, columns=['cid', 'Column Name', 'Data Type', 'Not Null', 'Default Value', 'Primary Key'])

    # Display the table name and its columns
    print(f"Table: {table}")
    display(table_info_df)
    print("\n")  # Add a newline for better readability
table_info_df
```

## Find pairs of patients treated by the same doctor.

```python
query = '''
SELECT P1.firstname||" "||p1.lastname as P1_Name, P2.firstname||" "||p2.lastname as P2_Name, c1.doctorid
from consultations c1
join consultations c2 on c1.doctorid = c2.doctorid and c1.patientid< c2.patientid
join patients p1 on c1.patientid = p1.patientid
join patients p2 on c2.patientid = p2.patientid
;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Show combinations of all doctors and hospitals.

```python
query = '''
SELECT d.firstname||" "||d.lastname as "Doctor_Name", h.Hospitalname
from doctors d
cross join hospitals h on d.hospitalid = h.hospitalid
order by h.Hospitalname;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## List all patients and their corresponding insurance plans.

```python
query = '''
SELECT p.firstname||" "||p.lastname as "Patient_Name", PlanName, InsuranceProvider
from patients p
inner join insuranceplans i on p.insuranceplanid = i.insuranceplanid;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Find patients who share the same insurance provider but live in different cities.

```python
query = '''
SELECT p1.PatientID AS Patient1_ID,
       p2.PatientID AS Patient2_ID,
       p1.InsurancePlanID
FROM Patients p1
JOIN Patients p2
  ON p1.InsurancePlanID = p2.InsurancePlanID
  AND p1.City <> p2.City
  AND p1.PatientID < p2.PatientID;

'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Find patients who have consulted the same doctor for different ailments.

```python
query = '''
SELECT c1.PatientID, c1.DoctorID, c1.ProcedureID, c2.ProcedureID
FROM Consultations c1
JOIN Consultations c2 ON c1.PatientID = c2.PatientID AND c1.DoctorID = c2.DoctorID AND c1.ProcedureID <> c2.ProcedureID;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

```python
query = '''
SELECT *
FROM consultations c1
join consultations c2 on c1.doctorid = c2.doctorid and c1.procedureid <> c2.procedureid
order by c1.doctorid, c2.doctorid;

'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## List all insurance claims along with patient and procedure details.

```python
query = '''
SELECT
   i.ClaimID,
   p.FirstName || ' ' || p.LastName AS PatientName,
   pr.ProcedureName, i.ClaimDate
FROM insuranceclaims i
join patients p on i.PatientID = p.PatientID
join procedures pr on i.procedureid = pr.procedureid;
'''
df = pd.read_sql_query(query, sqliteConnection)
print('The intention of the query here is to get all the insurance claims \nfor which there is information available in patient and procedure \ntable, we would have used left join if we wanted to revtrive the \nclaims even if there is no matching record available in other tables.\n\n')
df
```

## Identify doctors who treat patients from different cities but use the same procedure.

```python
query = '''
SELECT
    c1.DoctorID,
    c1.ProcedureID,
    p1.City AS City1,
    p2.City AS City2
FROM
    consultations c1
JOIN
    consultations c2
    ON c1.doctorid = c2.doctorid
    AND c1.procedureid = c2.procedureid
JOIN
    patients p1
    ON c1.patientid = p1.patientid
JOIN
    patients p2
    ON c2.patientid = p2.patientid
    AND p1.city <> p2.city
    AND p1.patientid < p2.patientid;

'''
df = pd.read_sql_query(query, sqliteConnection)
df
```
