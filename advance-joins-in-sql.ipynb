# Creating table

## Create table for Employees

```python
query = '''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    FirstName TEXT NOT NULL,
    LastName TEXT NOT NULL,
    DepartmentID INTEGER NOT NULL,
    ManagerID INTEGER,
    OfficeID INTEGER NOT NULL,
    Role TEXT NOT NULL,
    FOREIGN KEY (ManagerID) REFERENCES Employees(EmployeeID)
);
'''
execute(query)
```

## Create table for Projects

```python
query = '''
CREATE TABLE Projects (
    ProjectID INTEGER PRIMARY KEY,
    ProjectName TEXT NOT NULL,
    Description TEXT
);
'''
execute(query)
```

## Create table for EmployeeProjects

```python
query = '''
CREATE TABLE EmployeeProjects (
    EmployeeID INTEGER,
    ProjectID INTEGER,
    Role TEXT,
    PRIMARY KEY (EmployeeID, ProjectID),
    FOREIGN KEY (EmployeeID) REFERENCES Employees(EmployeeID),
    FOREIGN KEY (ProjectID) REFERENCES Projects(ProjectID)
);
'''
execute(query)
```

## Create table for Skills

```python
query = '''
CREATE TABLE Skills (
    SkillID INTEGER PRIMARY KEY,
    SkillName TEXT NOT NULL
);
'''
execute(query)
```

## Create table for EmployeeSkills

```python
query = '''
CREATE TABLE EmployeeSkills (
    EmployeeID INTEGER,
    SkillID INTEGER,
    PRIMARY KEY (EmployeeID, SkillID),
    FOREIGN KEY (EmployeeID) REFERENCES Employees(EmployeeID),
    FOREIGN KEY (SkillID) REFERENCES Skills(SkillID)
);
'''
execute(query)
```

## Create table for Offices

```python
query = '''
CREATE TABLE Offices (
    OfficeID INTEGER PRIMARY KEY,
    OfficeLocation TEXT NOT NULL
);
'''
execute(query)
```

## Create table for Categories

```python
query = '''
CREATE TABLE Categories (
    CategoryID INTEGER PRIMARY KEY,
    CategoryName TEXT NOT NULL
);
'''
execute(query)
```

## Create table for Products

```python
query = '''
CREATE TABLE Products (
    ProductID INTEGER PRIMARY KEY,
    ProductName TEXT NOT NULL,
    CategoryID INTEGER,
    FOREIGN KEY (CategoryID) REFERENCES Categories(CategoryID)
);
'''
execute(query)
```

## Find all pairs of employees who share the same department.

```python
query = '''
select e1.employeeID as employee1_ID, e1.firstName || ' ' || e1.LastName AS Employee1_Name,
       e2.EmployeeID as employee2_ID, e2.firstName || ' ' || e2.LastName AS Employee2_Name,
       e1.DepartmentID
from employees e1
join employees e2 ON e1.departmentID = e2.departmentID
where e1.employeeID < e2.employeeID and e1.departmentid = 1
order by Employee1_Name;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Show combinations of all employees and projects.

```python
query = '''
select e.EmployeeID,
       e.firstname || ' ' || e.lastname AS EmployeeName,
       p.projectID,
       p.projectname,
       ep.role
from employees e
cross join projects p
left join employeeprojects ep ON e.EmployeeID = ep.employeeID AND p.projectID = ep.projectID
order by EmployeeName;

'''
df = pd.read_sql_query(query, sqliteConnection)
print('The intention here is to see all possible combinations of employee and projects\n not to see who is assigned to which project, therefore a cross join has been used\n\n.')
df
```

## Identify employees who are their own managers.

```python
query = '''
SELECT e1.EmployeeID, e1.FirstName || ' ' || e1.LastName AS EmployeeName, e1.ManagerID
FROM employees e1
WHERE e1.EmployeeID = e1.ManagerID;
'''
df = pd.read_sql_query(query, sqliteConnection)
print('We do not have any such employee who are managing themselves.')
df
```

## Combine CROSS JOIN with filtering to match employees with specific skills.

```python
query = '''
SELECT *
FROM employees e cross join skills s
left join employeeskills es on e.employeeid = es.employeeid AND s.SkillID = es.SkillID
where es.skillid is not null;
'''
df = pd.read_sql_query(query, sqliteConnection)
print('The intention of the query is to match employees with their specific skills. \nInitially, a CROSS JOIN was used to generate all possible combinations of \nemployees and skills, followed by a LEFT JOIN to filter for actual skill matches. \nWhile the CROSS JOIN is unnecessary for efficiency as it produces redundant combinations, \nit has been retained in this small database to showcase the possibilities of using \nCROSS JOINs. A more efficient approach in larger datasets would involve directly joining \nthe employees and employeeskills tables to yield the desired results without excess data, \nthereby optimizing performance and enhancing clarity.\n\n')
df
```

## Write a query for project allocation.


Explaination: Create a query to display potential project assignments for pairs of employees who share the same manager. The query should show all possible project allocations, even if employees are not currently assigned to those projects.

```python
query = '''
select
  e1.employeeid as EmployeeID,
  e1.firstname||' '||e1.lastname as E1_EmployeeName,
  e2.employeeid as EmployeeID,
  e2.firstname||' '||e2.lastname as E2_EmployeeName,
  p.projectid,
  p.projectname,
  ep1.role as Employee1_Role,
  ep2.role as Employee2_role
from employees e1
join employees e2 on e1.ManagerID = e2.ManagerID -- self join for same manager
cross join projects p
left join employeeprojects ep1 on e1.EmployeeID = ep1.EmployeeID and p.projectid = ep1.projectid
left join employeeprojects ep2 on e2.EmployeeID = ep2.EmployeeID and p.projectid = ep2.projectid
where e1.EmployeeID < e2.EmployeeID
order by projectname and Employee1_Role;
'''
df = pd.read_sql_query(query, sqliteConnection)
print('This query allows you to view potential project allocations, showing current roles or null \nvalues where employees have no project assignment. It leverages both self joins and cross joins \nto provide a full view of possible employee-project assignments based on shared managers.\n\n')
df
```

## List all employee pairs who work in the same office but report to different managers.

```python
query = '''
select
  e1.employeeid as EmployeeID,
  e1.firstname||' '||e1.lastname as E1_EmployeeName,
  e1.officeid,
  e1.managerid,
  e2.employeeid as EmployeeID,
  e2.firstname||' '||e2.lastname as E2_EmployeeName,
  e2.officeid,
  e2.managerid
from employees e1
join employees e2 on e1.officeID = e2.officeID -- self join for same office
where e1.EmployeeID < e2.EmployeeID and e1.managerid <> e2.managerid
;

'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Show all possible team configurations for a specific project

```python
query = '''
select
  e1.employeeid,
  e1.firstname||' '||e1.lastname as EmployeeName,
  e2.employeeid,
  e2.firstname||' '||e2.lastname as EmployeeName,
  p.projectid,
  p.projectname
from employees e1
cross join employees e2
join employeeprojects ep1 on e1.employeeid = ep1.employeeid
join employeeprojects ep2 on e2.employeeid = ep2.employeeid
join projects p on ep1.projectid = p.projectid and ep2.projectid = p.projectid
where e1.employeeid < e2.employeeid;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```

## Find employees who have worked on the same project as themselves but in different roles.

```python
query = '''
SELECT
    ep1.EmployeeID,
    e.FirstName || ' ' || e.LastName AS EmployeeName,
    ep1.ProjectID,
    p.ProjectName,
    ep1.Role AS Role1,
    ep2.Role AS Role2
FROM EmployeeProjects ep1
join employeeprojects ep2
on ep1.employeeid = ep2.employeeid
and ep1.projectid = ep2.projectid
and ep1.role <> ep2.role
join projects p
on ep1.projectid = p.projectid
join employees e
on ep1.employeeid = e.employeeid
;
'''
df = pd.read_sql_query(query, sqliteConnection)
df
```
